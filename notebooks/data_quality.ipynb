**¿Qué diferencias existen entre int64, float64 y object?**
`int64` guarda números enteros (como identificadores o la hora del día). `float64` guarda números decimales; Pandas también lo usa cuando una columna de enteros tiene valores vacíos (NaN), porque NaN no se puede representar como entero. `object` guarda texto, como los nombres de productos, pasillos y departamentos.

**¿Qué información proporciona info() que no se obtiene directamente con shape?**
`shape` solo dice cuántas filas y columnas hay. `info()` además muestra el nombre de cada columna, su tipo de dato, cuántos valores no nulos tiene cada una y la memoria que ocupa el DataFrame. Gracias a eso se pueden detectar datos faltantes.

- `product_name` en products: 1.258 nulos.

## Momento 4. Indexación y filtrado lógico de la información

```python
ej1 = products[["product_id", "product_name"]]
print("Registros:", ej1.shape)
display(ej1.head())
```

**Ejercicio 1.** Se seleccionaron solo las columnas `product_id` y `product_name` de products. El resultado conserva las 49.694 filas, porque seleccionar columnas no elimina registros: solo reduce la información que se muestra a lo que se necesita.

```python
ej2 = products.loc[products["department_id"] == 1]
print("Registros:", ej2.shape)
display(ej2.head())
```

**Ejercicio 2.** Hay 4.008 productos en el departamento 1 (frozen, congelados), que equivalen a cerca del 8,1% del catálogo.

```python
ej3 = orders.loc[orders["order_hour_of_day"] == 10]
print("Registros:", ej3.shape)
display(ej3.head())
```

**Ejercicio 3.** Se realizaron 40.578 pedidos a las 10 horas, aproximadamente el 8,5% del total de pedidos. Esto indica que media mañana es un horario con actividad de compra importante.

```python
ej4 = orders.loc[orders["order_dow"] == 0]
print("Registros:", ej4.shape)
display(ej4.head())
```

**Ejercicio 4.** Hay 84.090 pedidos con `order_dow` igual a 0, cerca del 17,6% del total. Si los 7 días tuvieran el mismo peso, cada uno tendría alrededor del 14,3%, así que el día 0 tiene más pedidos que el promedio. En este dataset el valor 0 suele interpretarse como domingo.

```python
ej5 = orders.loc[orders["order_number"] > 10]
print("Registros:", ej5.shape)
display(ej5.head())
```

**Ejercicio 5.** 244.072 pedidos tienen `order_number` mayor a 10, es decir, aproximadamente el 51% de los pedidos son del pedido número 11 en adelante de un cliente. Esto muestra que muchos registros provienen de clientes recurrentes.

```python
ej6 = orders.loc[orders["days_since_prior_order"] > 15]
print("Registros:", ej6.shape)
display(ej6.head())
```

**Ejercicio 6.** 110.706 pedidos se hicieron más de 15 días después del pedido anterior, cerca del 23% del total. Los 28.819 pedidos con valor nulo no entran en el filtro, porque una comparación con NaN siempre da falso.

```python
ej7 = order_products.loc[order_products["reordered"] == 1]
print("Registros:", ej7.shape)
display(ej7.head())
```

**Ejercicio 7.** 2.683.838 artículos tienen `reordered` igual a 1, alrededor del 59% de los registros de order_products. Es decir, la mayoría de los productos comprados ya habían sido pedidos antes por el mismo cliente.

```python
ej8 = order_products.loc[order_products["add_to_cart_order"] == 1]
print("Registros:", ej8.shape)
display(ej8.head())
```

**Ejercicio 8.** Hay 450.046 registros con `add_to_cart_order` igual a 1. Como cada pedido tiene un solo producto en la primera posición, este número se aproxima a la cantidad de pedidos que aparecen en order_products. Es menor que los 478.967 pedidos de orders, lo que sugiere que no todos los pedidos tienen su detalle de productos registrado o que algunos tienen la posición vacía (836 nulos).

```python
filtro1 = orders.loc[(orders["order_hour_of_day"] == 10) & (orders["order_number"] > 10)]
print("Registros:", filtro1.shape)
display(filtro1.head())
```

**Filtro combinado 1.** Hay 21.537 pedidos realizados a las 10 horas por clientes con más de 10 pedidos. El resultado es menor que cada filtro por separado (40.578 a las 10 horas y 244.072 con order_number > 10), porque un registro debe cumplir ambas condiciones. Aproximadamente el 53% de los pedidos de las 10 horas corresponden a clientes frecuentes.

```python
filtro2 = products.loc[(products["department_id"] == 1) & (products["aisle_id"] > 50)]
print("Registros:", filtro2.shape)
display(filtro2.head())
```

**Filtro combinado 2.** Hay 1.618 productos del departamento 1 con `aisle_id` mayor a 50, cerca del 40% de los 4.008 productos de ese departamento. Al exigir las dos condiciones, el resultado se reduce.

```python
resumen = {
    "Ej1 - product_id y product_name": ej1.shape[0],
    "Ej2 - department_id == 1": ej2.shape[0],
    "Ej3 - order_hour_of_day == 10": ej3.shape[0],
    "Ej4 - order_dow == 0": ej4.shape[0],
    "Ej5 - order_number > 10": ej5.shape[0],
    "Ej6 - days_since_prior_order > 15": ej6.shape[0],
    "Ej7 - reordered == 1": ej7.shape[0],
    "Ej8 - add_to_cart_order == 1": ej8.shape[0],
    "Filtro 1 - hora 10 y order_number > 10": filtro1.shape[0],
    "Filtro 2 - depto 1 y aisle_id > 50": filtro2.shape[0],
}
for k, v in resumen.items():
    print(f"{k}: {v:,}")
```

### Respuestas del Momento 4

**¿Qué diferencia existe entre seleccionar columnas y filtrar filas?**
Seleccionar columnas reduce las variables que se muestran pero mantiene todos los registros (ejercicio 1: siguen siendo 49.694 filas). Filtrar filas conserva las columnas pero deja solo los registros que cumplen una condición (ejercicio 2: de 49.694 productos quedan 4.008).

**¿Qué ventajas ofrece loc[] al consultar información específica?**
`loc[]` permite filtrar filas por condiciones y seleccionar columnas por su nombre en una sola instrucción, por ejemplo `df.loc[condición, ["col1", "col2"]]`. Hace el código más claro y legible, y evita ambigüedades al trabajar con etiquetas en lugar de posiciones.

**¿Qué ocurre con la cantidad de registros cuando se aplican varias condiciones?**
Disminuye, porque con `&` cada registro debe cumplir todas las condiciones al mismo tiempo. Por ejemplo, de 40.578 pedidos a las 10 horas, solo 21.537 son además de clientes con más de 10 pedidos.

**¿Por qué es importante utilizar correctamente los operadores lógicos?**
Porque cambian completamente el resultado: `&` (y) exige que se cumplan todas las condiciones y reduce los registros, mientras que `|` (o) acepta cualquiera y los aumenta. Además, en Pandas cada condición debe ir entre paréntesis y no se pueden usar `and`/`or`; si se hace mal, el código da error o devuelve datos que no responden la pregunta.
